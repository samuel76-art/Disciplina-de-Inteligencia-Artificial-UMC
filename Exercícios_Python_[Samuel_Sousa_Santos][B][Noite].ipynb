```python
from ast import Break
#Exercicio 1
print("hello word")

#Execicio 2
numero1 = float(input("Digite um número:"))
numero2 = float(input("Digite um número:"))
soma = numero1 + numero2
print("A soma dos números é:",soma)

#Exercicio 3
raio = float(input("Digite o raio: "))
area = 3.14 * raio ** 2
print("A area é:", area)

#Exercicio 4
celcius = float(input("Digite a temperatura em fahrenheit: "))
fahrenheit = celcius * 9/5 +32
print("A temperatura em celcius é: ",fahrenheit)

#Exercicio 5
numero = int(input("Digite um numero: "))
for i in range(1,11):
  print(numero, "x" , i, "=", numero * i)

#Exercicio 6
print("Os números pares são:")
for numero in range(1,21):
  if numero % 2 == 0:
    print(numero, end=" ")

#Exercicio 7
numeroPar = int(input("Digite um número: "))
if numeroPar %2 == 0:
  print("O numero é par")
else:
  print("O numero é impar")

#Exercicio 8
palavra = input("Digite uma palavra: ")
palavra = palavra.lower()
if palavra == palavra[::-1]:
  print("A palavra é um palíndromo")
else:
  print("A palavra não é um palíndromo")

#Exercicio 9
numero = int(input("Digite um número: "))
fatorial = 1
for i in range(1, numero + 1):
  fatorial = fatorial * 1
print("O fatorial de", numero, "é", fatorial)

#Exercicio 10
n1 = 0
n2 = 1
print("Sequencia de Fibonacci: ")
for i in range(10):
  print(n1)
  n1, n2 = n2, n1 + n2

#Exercicio 11
numeros = []
quantidade = int(input("Quantos numeros voce deseja digitar: "))
for i in range(quantidade):
  numero = float(input("Digite um numero: "))
  numeros.append(numero)
maior = max(numeros)
menor = min(numeros)
print("A lista de numeros é: ",numeros)
print("O maior numero é: ",maior)
print("O menor numero é: ",menor)

#Exercicio 12
texto = input("Digite um texto: ")
contagem = {}
for caractere in texto:
  if caractere in contagem:
    contagem[caractere] += 1
  else:
    contagem[caractere] = 1
print("Contagem de caractere: ")
for caractere, quantidade in contagem.items():
  print(f"'{caractere}': {quantidade}")

#Exercicio 13
numeros = []
quantidade = int(input("Quantos números você quer digitar? "))
for i in range(quantidade):
  numero = float(input("Digite o numero: "))
  numeros.append(numero)
numeros.sort()
print("A lista em ordem crescente é: ", numeros)

#Exercicio 14
numeros = []
quantidade = int(input("Quantos números você quer digitar? "))
for i in range(quantidade):
    numero = float(input("Digite um número: "))
    numeros.append(numero)
soma = sum(numeros)
print("A soma de todos os elementos é:", soma)

#Exercicio 15
numeros = []
quantidade = int(input("Quantos números você quer digitar? "))
for i in range(quantidade):
    numero = float(input("Digite um número: "))
    numeros.append(numero)
soma = sum(numeros)
media = soma / len(numeros)
print("A média dos numeros é: ", media)

#Exercicio 16
texto = input("Digite um texto: ")
textoInvertido = texto [::-1]
print("Seu texto invertido é: ", textoInvertido)

#Exercicio 17
texto = input("Digite seu texto: ")
palavras = texto.split()
contagem = {}
for palavra in palavras:
  if palavra in contagem:
    contagem[palavra] += 1
  else:
    contagem[palavra] = 1
    print("A frequencia de cada palavra é: ")
    for palavra, quantidade in contagem.items():
      print(f"'{palavra}': {quantidade}")

#Exercicio 18
numero = int(input("Digite um numero: "))
if numero < 2:
    print(numero, "não é primo")
else:
    primo = True
    for i in range(2, int(numero ** 0.5) + 1):
        if numero % i == 0:
            primo = False
            break
    if primo:
        print(numero, "é primo")
    else:
        print(numero, "não é primo")

#Exercicio 19
import random
numeroSecreto = random.randint(1, 10)
print("Adivinhe o número entre 1 e 10")
chute = int(input("Digite seu chute: "))
while chute != numeroSecreto:
    if chute < numeroSecreto:
        print("Muito baixo. Tente novamente")
    else:
        print("Muito alto. Tente novamente.")
    chute = int(input("Digite seu chute: "))
print("Parabéns. Você acertou o número", numeroSecreto)

#Exercicio 20
linhas = int(input("Quantas linhas a matriz vai ter? "))
colunas = int(input("Quantas colunas a matriz vai ter? "))
matriz = []
for i in range(linhas):
    nova_linha = []
    for j in range(colunas):
        valor = float(input(f"Digite o valor da posição [{i}][{j}]: "))
        nova_linha.append(valor)
    matriz.append(nova_linha)
transposta = []
for coluna in range(len(matriz[0])):
    nova_linha = []
    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])
    transposta.append(nova_linha)
print("Matriz original:")
for linha in matriz:
    print(linha)
print("Matriz transposta:")
for linha in transposta:
    print(linha)

#Exercicio 21
def buscaBinaria(lista, alvo):
  esquerda, direita = 0, len(lista) - 1
  while esquerda <= direita:
    meio = (esquerda + direita) // 2
    if lista[meio] == alvo:
      return meio
    elif lista[meio] < alvo:
      esquerda = meio + 1
    else:
      direita = meio - 1
  return -1
quantidade = int(input("Quantos numeros você quer digitar: "))
lista = []
for i in range(quantidade):
  numero = int(input(f"Digite um numero {i + 1}: "))
  lista.append(numero)
lista.sort()
print(f"\nLista ordenada: {lista}")
alvo = int(input("\nQual numero deseja procurar? "))
resultado = buscaBinaria(lista, alvo)
if resultado != -1:
  print(f"\nO numero {alvo} foi encontrado no indice {resultado}")
else:
  print(f"\nO numero {alvo} não está na lista")

#exercicio 22
def CifraDeCesar(mensagem, deslocamento):
  resultado = " "
  for caractere in mensagem:
    if caractere.isalpha():
      base = ord('A') if caractere.isupper() else ord('a')
      posicao = (ord(caractere) - base + deslocamento) % 26
      novaLetra = chr(base + posicao)
      resultado += novaLetra
    else:
      resultado += caractere
  return resultado
mensagem = input("Digite uma mensagem para ser criptografada: ")
deslocamento = int(input("Digite o deslocamento: "))
mensagemCriptografada = CifraDeCesar(mensagem, deslocamento)
print(f"\nMensagem criptografada é: {mensagemCriptografada}")

#exercicio 23
def decifraCesar(mensagem, deslocamento):
  resultado = ""
  for caractere in mensagem:
    if caractere.isalpha():
       base = ord('A') if caractere.isupper() else ord('a')
       posicao = (ord(caractere) - base - deslocamento) % 26
       novaLetra = chr(base + posicao)
       resultado += novaLetra
    else:
      resultado += caractere
  return resultado
textoCriptografado = input("Digite o texto que deseja descriptografar: ")
deslocamento = int(input("Digite o deslocamento de casas: "))
mensagemOriginal = decifraCesar(textoCriptografado, deslocamento)
print(f"\nA mensagem original e: {mensagemOriginal}")

#exercicio 24
def validarCPF(cpf):
  cpf = ''.join(filter(str.isdigit, cpf))
  if len(cpf) != 11:
    return False
  if cpf == cpf[0] * 11:
    return False
  soma = 0
  for i in range (9):
    soma += int(cpf[i]) * (10 - i)
  resto = soma % 11
  digito1 = 0 if resto < 2 else 11 - resto
  if digito1 != int(cpf[9]):
    return False
  soma = 0
  for i in range(10):
    soma += int(cpf[i]) * (11 - i)
  resto = soma % 11
  digito2 = 0 if resto < 2 else 11 - resto
  if digito2 != int(cpf[10]):
    return False
  return True
cpfDigitado = input("Digite o CPF: ")
if validarCPF(cpfDigitado):
    print("\nCPF válido")
else:
    print("\nCPF inválido")

#exercicio 25
def classificarNumeros(lista):
  positivos = []
  negativos = []
  zeros = []
  for numero in lista:
    if numero > 0:
      positivos.append(numero)
    elif numero < 0:
      negativos.append(numero)
    else:
      zeros.append(numero)
  return positivos, negativos, zeros
quantidade = int(input("quantos numeros voce deseja digitar: "))
lista = []
for i in range(quantidade):
  numero = float(input(f"Digite o numero {i + 1}: "))
  lista.append(numero)
positivos, negativos, zero = classificarNumeros(lista)
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"zeros: {zero}")

#Exercicio 26
def somar(a, b):
    return a + b
def subtrair(a, b):
    return a - b
def multiplicar(a, b):
    return a * b
def dividir(a, b):
    if b == 0:
        return "Erro: divisão por zero não é permitida!"
    return a / b
print("=== Calculadora ===")
num1 = float(input("Digite o primeiro número: "))
operador = input("Digite a operação (+, -, *, /): ")
num2 = float(input("Digite o segundo número: "))
if operador == "+":
    resultado = somar(num1, num2)
elif operador == "-":
    resultado = subtrair(num1, num2)
elif operador == "*":
    resultado = multiplicar(num1, num2)
elif operador == "/":
    resultado = dividir(num1, num2)
else:
    resultado = "Operador inválido"
print(f"\nResultado: {resultado}")

#exercicio 27
def sao_anagramas(palavra1, palavra2):
    palavra1 = palavra1.replace(" ", "").lower()
    palavra2 = palavra2.replace(" ", "").lower()
    return sorted(palavra1) == sorted(palavra2)
palavra1 = input("Digite a primeira palavra: ")
palavra2 = input("Digite a segunda palavra: ")
if sao_anagramas(palavra1, palavra2):
    print(f"\n'{palavra1}' e '{palavra2}' são anagramas")
else:
    print(f"\n'{palavra1}' e '{palavra2}' não são anagramas.")
#exercicio 28
import matplotlib.pyplot as plt
quantidade = int(input("Quantas categorias você quer no gráfico? "))
categorias = []
valores = []
for i in range(quantidade):
    nome = input(f"Digite o nome da categoria {i + 1}: ")
    valor = float(input(f"Digite o valor para '{nome}': "))
    categorias.append(nome)
    valores.append(valor)
plt.bar(categorias, valores, color='skyblue')
plt.title("Gráfico de Barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()

#exercicio 29
inicio = int(input("Digite o número inicial: "))
fim = int(input("Digite o número final: "))
for numero in range(inicio, fim + 1):
    if numero % 3 == 0 and numero % 5 == 0:
        print("FizzBuzz")
    elif numero % 3 == 0:
        print("Fizz")
    elif numero % 5 == 0:
        print("Buzz")
    else:
        print(numero)

#exercicio 30
import random
def escolherPalavra():
    palavras = ["gato", "casa", "bola", "sol", "mesa", "livro"]
    return random.choice(palavras)
def jogarForca():
    palavra = escolherPalavra()
    letrasDescobertas = ["_"] * len(palavra)
    letrasTentadas = []
    tentativasRestantes = 6
    print("=== Jogo da Forca ===")
    print(f"A palavra tem {len(palavra)} letras.")
    while tentativasRestantes > 0 and "_" in letrasDescobertas:
        print(f"\nPalavra: {' '.join(letrasDescobertas)}")
        print(f"Tentativas restantes: {tentativasRestantes}")
        print(f"Letras já tentadas: {', '.join(letrasTentadas) if letrasTentadas else 'nenhuma'}")
        letra = input("Digite uma letra: ").lower()
        if len(letra) != 1 or not letra.isalpha():
            print("Digite apenas uma letra válida")
            continue
        if letra in letrasTentadas:
            print("Você já tentou essa letra!")
            continue
        letrasTentadas.append(letra)
        if letra in palavra:
            print("Boa Letra correta.")
            for i, char in enumerate(palavra):
                if char == letra:
                    letrasDescobertas[i] = letra
        else:
            tentativasRestantes -= 1
            print(f"Letra errada Restam {tentativasRestantes} tentativas.")

    if "_" not in letrasDescobertas:
        print(f"\nParabéns Você acertou a palavra: {palavra}")
    else:
        print(f"\nVocê perdeu A palavra era: {palavra}")
jogarForca()

#exercicio 31
def bubbleSort(lista):
    n = len(lista)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
print(f"\nLista original: {lista}")
listaOrdenada = bubbleSort(lista)
print(f"Lista ordenada: {listaOrdenada}")

#exercicio 32
def criarTabuleiro():
    return [" " for _ in range(9)]
def mostrarTabuleiro(tabuleiro):
    print(f"\n {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
def verificarVencedor(tabuleiro):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]
    for a, b, c in combinacoes:
        if tabuleiro[a] != " " and tabuleiro[a] == tabuleiro[b] == tabuleiro[c]:
            return tabuleiro[a]
    return None
def tabuleiroCheio(tabuleiro):
    return " " not in tabuleiro
def jogarVelha():
    tabuleiro = criarTabuleiro()
    jogadorAtual = "X"
    print("=== Jogo da Velha ===")
    print("Posições do tabuleiro:")
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")
    while True:
        mostrarTabuleiro(tabuleiro)
        print(f"\nVez do jogador {jogadorAtual}")
        posicao = input("Escolha uma posição (1-9): ")
        if not posicao.isdigit() or int(posicao) < 1 or int(posicao) > 9:
            print("Posição inválida! Digite um número de 1 a 9.")
            continue
        posicao = int(posicao) - 1
        if tabuleiro[posicao] != " ":
            print("Essa posição já está ocupada")
            continue
        tabuleiro[posicao] = jogadorAtual
        vencedor = verificarVencedor(tabuleiro)
        if vencedor:
            mostrarTabuleiro(tabuleiro)
            print(f"\nO jogador {vencedor} venceu")
            break
        if tabuleiroCheio(tabuleiro):
            mostrarTabuleiro(tabuleiro)
            print("\nEmpate")
            break
        jogadorAtual = "O" if jogadorAtual == "X" else "X"
jogarVelha()

#exercicio 33
def sao_anagramas(palavra1, palavra2):
    palavra1 = palavra1.replace(" ", "").lower()
    palavra2 = palavra2.replace(" ", "").lower()
    return sorted(palavra1) == sorted(palavra2)
palavra1 = input("Digite a primeira palavra: ")
palavra2 = input("Digite a segunda palavra: ")
if sao_anagramas(palavra1, palavra2):
    print(f"\n'{palavra1}' e '{palavra2}' são anagramas")
else:
    print(f"\n'{palavra1}' e '{palavra2}' não são anagramas.")

#exercicio 34
def contarVogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for caractere in texto:
        if caractere in vogais:
            contador += 1
    return contador
texto = input("Digite uma frase ou palavra: ")
totalVogais = contarVogais(texto)
print(f"\nA string '{texto}' tem {totalVogais} vogal(is).")

#exercicio 35
def fibonacciNTermos(n):
    sequencia = []
    a, b = 0, 1
    for _ in range(n):
        sequencia.append(a)
        a, b = b, a + b
    return sequencia
n = int(input("Quantos termos da sequência de Fibonacci você quer gerar? "))
resultado = fibonacciNTermos(n)
print(f"\nOs primeiros {n} termos de Fibonacci: {resultado}")

#exercicio 36
import random
import string
def gerarSenha(tamanho):
    caracteres = string.ascii_letters + string.digits
    senha = ""
    for _ in range(tamanho):
        senha += random.choice(caracteres)
    return senha
tamanho = int(input("Quantos caracteres a senha deve ter? "))
senhaGerada = gerarSenha(tamanho)
print(f"\nSenha gerada: {senhaGerada}")

#exercicio 37
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    return n * fatorial(n - 1)
numero = int(input("Digite um número para calcular o fatorial: "))
resultado = fatorial(numero)
print(f"\nO fatorial de {numero} é {resultado}")

#exercicio 38
def contarPalavras(texto):
    palavras = texto.split()
    return len(palavras)
texto = input("Digite um texto: ")
totalPalavras = contarPalavras(texto)
print(f"\nO texto tem {totalPalavras} palavra(s).")

#exercicio 39
import random
def lancarDados():
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)
    return dado1, dado2
dado1, dado2 = lancarDados()
print(f"\nDado 1: {dado1}")
print(f"Dado 2: {dado2}")
print(f"Soma: {dado1 + dado2}")

#exercicio 40
def adicionarItem(lista, item):
    lista.append(item)
    print(f"\n'{item}' foi adicionado à lista.")
def removerItem(lista, item):
    if item in lista:
        lista.remove(item)
        print(f"\n'{item}' foi removido da lista.")
    else:
        print(f"\n'{item}' não está na lista.")
def listarItens(lista):
    if not lista:
        print("\nA lista de compras está vazia.")
    else:
        print("\nLista de compras:")
        for i, item in enumerate(lista, 1):
            print(f"{i}. {item}")
listaCompras = []
while True:
    print("\n1. Adicionar item")
    print("2. Remover item")
    print("3. Listar itens")
    print("4. Sair")
    opcao = input("Escolha uma opção: ")
    if opcao == "1":
        item = input("Digite o item a adicionar: ")
        adicionarItem(listaCompras, item)
    elif opcao == "2":
        item = input("Digite o item a remover: ")
        removerItem(listaCompras, item)
    elif opcao == "3":
        listarItens(listaCompras)
    elif opcao == "4":
        print("\nAté logo!")
        break
    else:
        print("\nOpção inválida!")

 #exercicio 41
```
