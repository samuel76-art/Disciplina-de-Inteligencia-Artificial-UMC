```python
#Exercicio 18 - Numero primo
#Objetivo - Verificar se um numero inteiro fornecido pelo usuario e primo
#Raciocinio - Peco um numero ao usuario e verifico se e menor que 2 pois nao e primo se for maior crio uma variavel primo como true e uso o for ate a raiz quadrada do numero para ver se ele tem algum divisor exato com o operador % se achar defino primo como false e paro o laço com break mostrando o resultado no final
numero = int(input("Digite um numero: "))
if numero < 2:
    print(numero, "não é primo")
else:
    primo = True
    for i in range(2, int(numero ** 0.5) + 1):
        if numero % i == 0:
            primo = False
            break
    if primo:
        print(numero, "é primo")
    else:
        print(numero, "não é primo")
```

```python
#Exercicio 20 - Matriz transposta
#Objetivo - Ler os valores de uma matriz e construir a sua matriz transposta trocando linhas por colunas
#Raciocinio - Peco as dimensoes de linhas e colunas e uso dois for aninhados para preencher a matriz original depois crio outra estrutura com for trocando os indices de linha e coluna para gerar a transposta e exibo as duas no final
linhas = int(input("Quantas linhas a matriz vai ter? "))
colunas = int(input("Quantas colunas a matriz vai ter? "))
matriz = []
for i in range(linhas):
    nova_linha = []
    for j in range(colunas):
        valor = float(input(f"Digite o valor da posição [{i}][{j}]: "))
        nova_linha.append(valor)
    matriz.append(nova_linha)
transposta = []
for coluna in range(len(matriz[0])):
    nova_linha = []
    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])
    transposta.append(nova_linha)
print("Matriz original:")
for linha in matriz:
    print(linha)
print("Matriz transposta:")
for linha in transposta:
    print(linha)
```

```python
#exercicio 22 -  Criptografia de Cifra de César
#Objetivo - Criptografar uma mensagem deslocando as letras do alfabeto de acordo com um numero fornecido pelo usuario
#Raciocinio - Crio a funcao cifradecesar que percorre cada caractere da mensagem se for letra descubro a base ascii maiuscula ou minuscula e calculo a nova posicao usando % 26 para dar a volta no alfabeto e reconstruo a string mantendo espacos e simbolos intactos
def CifraDeCesar(mensagem, deslocamento):
  resultado = ""
  for caractere in mensagem:
    if caractere.isalpha():
      base = ord('A') if caractere.isupper() else ord('a')
      posicao = (ord(caractere) - base + deslocamento) % 26
      novaLetra = chr(base + posicao)
      resultado += novaLetra
    else:
      resultado += caractere
  return resultado
mensagem = input("Digite uma mensagem para ser criptografada: ")
deslocamento = int(input("Digite o deslocamento: "))
mensagemCriptografada = CifraDeCesar(mensagem, deslocamento)
print(f"\nMensagem criptografada é: {mensagemCriptografada}")
```

```python
#Exercicio 23 - Descriptografar Cifra de César
#Objetivo - Descriptografar uma mensagem deslocando as letras do alfabeto no sentido inverso de acordo com um numero fornecido pelo usuario
#Raciocinio - Crio a funcao decifrasesar que percorre cada caractere da mensagem e se for letra descubro a base ascii maiuscula ou minuscula e subtraio o deslocamento usando % 26 para voltar a posicao original reconstruindo a mensagem e mantendo espacos e simbolos intactos
def decifraCesar(mensagem, deslocamento):
  resultado = ""
  for caractere in mensagem:
    if caractere.isalpha():
       base = ord('A') if caractere.isupper() else ord('a')
       posicao = (ord(caractere) - base - deslocamento) % 26
       novaLetra = chr(base + posicao)
       resultado += novaLetra
    else:
      resultado += caractere
  return resultado
textoCriptografado = input("Digite o texto que deseja descriptografar: ")
deslocamento = int(input("Digite o deslocamento de casas: "))
mensagemOriginal = decifraCesar(textoCriptografado, deslocamento)
print(f"\nA mensagem original e: {mensagemOriginal}")
```

```python
#exercicio 24 - Validacao do CPF
#Objetivo - Verificar se um numero de cpf digitado pelo usuario e valido calculando os dois digitos verificadores
#Raciocinio - Limpo o cpf deixando so numeros e valido se tem 11 digitos ou se e uma sequencia repetida depois calculo a soma multiplicando os 9 primeiros digitos por pesos regressivos para achar o primeiro digito verificador e repito a logica com os 10 primeiros para achar o segundo confirmando se bate com o informado
def validarCPF(cpf):
  cpf = ''.join(filter(str.isdigit, cpf))
  if len(cpf) != 11:
    return False
  if cpf == cpf[0] * 11:
    return False
  soma = 0
  for i in range (9):
    soma += int(cpf[i]) * (10 - i)
  resto = soma % 11
  digito1 = 0 if resto < 2 else 11 - resto
  if digito1 != int(cpf[9]):
    return False
  soma = 0
  for i in range(10):
    soma += int(cpf[i]) * (11 - i)
  resto = soma % 11
  digito2 = 0 if resto < 2 else 11 - resto
  if digito2 != int(cpf[10]):
    return False
  return True
cpfDigitado = input("Digite o CPF: ")
if validarCPF(cpfDigitado):
    print("\nCPF válido")
else:
    print("\nCPF inválido")
```

```python
#Exercicio 25 - Classifcacao de numeros
#Objetivo - Separar uma lista de numeros informada pelo usuario em tres categorias positivos negativos e zeros
#Raciocinio - Peco os numeros ao usuario e crio a funcao classificarnumeros que percorre a lista usando um if para testar se cada valor e maior menor ou igual a zero salvando em listas separadas e retornando as tres no final para exibicao
def classificarNumeros(lista):
  positivos = []
  negativos = []
  zeros = []
  for numero in lista:
    if numero > 0:
      positivos.append(numero)
    elif numero < 0:
      negativos.append(numero)
    else:
      zeros.append(numero)
  return positivos, negativos, zeros
quantidade = int(input("quantos numeros voce deseja digitar: "))
lista = []
for i in range(quantidade):
  numero = float(input(f"Digite o numero {i + 1}: "))
  lista.append(numero)
positivos, negativos, zeros = classificarNumeros(lista)
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Zeros: {zeros} (total: {len(zeros)})")
```

```python
#Exercicio 26 - Calculadora simples
#Objetivo - Realizar operacoes matematicas basicas entre dois numeros informados pelo usuario com base no operador escolhido
#Raciocinio - Crio funcoes para cada operacao somar subtrair multiplicar e dividir tratando a divisao por zero depois peco dois numeros e o operador ao usuario uso o if para chamar a funcao correspondente e exibo o resultado no final
def somar(a, b):
    return a + b
def subtrair(a, b):
    return a - b
def multiplicar(a, b):
    return a * b
def dividir(a, b):
    if b == 0:
        return "Erro: divisão por zero não é permitida!"
    return a / b
print("=== Calculadora ===")
num1 = float(input("Digite o primeiro número: "))
operador = input("Digite a operação (+, -, *, /): ")
num2 = float(input("Digite o segundo número: "))
if operador == "+":
    resultado = somar(num1, num2)
elif operador == "-":
    resultado = subtrair(num1, num2)
elif operador == "*":
    resultado = multiplicar(num1, num2)
elif operador == "/":
    resultado = dividir(num1, num2)
else:
    resultado = "Operador inválido"
print(f"\nResultado: {resultado}")
```

```python
#exercicio 27 - Anagramas
#Objetivo - Verificar se duas palavras ou frases informadas pelo usuario sao anagramas entre si
#Raciocinio - Crio a funcao sao_anagramas que limpa os espacos e transforma o texto para minusculo e compara as duas strings ordenadas com sorted se a lista de caracteres for igual retorna true depois peco as palavras e exibo o resultado no print
def sao_anagramas(palavra1, palavra2):
    palavra1 = palavra1.replace(" ", "").lower()
    palavra2 = palavra2.replace(" ", "").lower()
    return sorted(palavra1) == sorted(palavra2)
palavra1 = input("Digite a primeira palavra: ")
palavra2 = input("Digite a segunda palavra: ")
if sao_anagramas(palavra1, palavra2):
    print(f"\n'{palavra1}' e '{palavra2}' são anagramas")
else:
    print(f"\n'{palavra1}' e '{palavra2}' não são anagramas.")
```

```python
#Exercicio 30 - Jogo da forca
#Objetivo - Criar um jogo da forca interativo onde o usuario tenta adivinhar uma palavra sorteada letra por letra com um limite de tentativas
#Raciocinio - Importo a biblioteca random para escolher uma palavra aleatoria de uma lista predefinida crio uma lista com underline para representar as letras ocultas e controlo as tentativas restantes e letras ja chutadas usando um laco while que valida se a letra e valida e se ja foi digitada atualizando o painel do jogo ate a palavra ser descoberta ou as chances acabarem
import random
random.seed()
def escolherPalavra():
    palavras = ["gato", "casa", "bola", "sol", "mesa", "livro"]
    return random.choice(palavras)
def jogarForca():
    palavra = escolherPalavra()
    letrasDescobertas = ["_"] * len(palavra)
    letrasTentadas = []
    tentativasRestantes = 6
    print("=== Jogo da Forca ===")
    print(f"A palavra tem {len(palavra)} letras.")
    while tentativasRestantes > 0 and "_" in letrasDescobertas:
        print(f"\nPalavra: {' '.join(letrasDescobertas)}")
        print(f"Tentativas restantes: {tentativasRestantes}")
        print(f"Letras já tentadas: {', '.join(letrasTentadas) if letrasTentadas else 'nenhuma'}")
        letra = input("Digite uma letra: ").lower()
        if len(letra) != 1 or not letra.isalpha():
            print("Digite apenas uma letra válida")
            continue
        if letra in letrasTentadas:
            print("Você já tentou essa letra!")
            continue
        letrasTentadas.append(letra)
        if letra in palavra:
            print("Boa Letra correta.")
            for i, char in enumerate(palavra):
                if char == letra:
                    letrasDescobertas[i] = letra
        else:
            tentativasRestantes -= 1
            print(f"Letra errada Restam {tentativasRestantes} tentativas.")
    if "_" not in letrasDescobertas:
        print(f"\nParabéns Você acertou a palavra: {palavra}")
    else:
        print(f"\nVocê perdeu A palavra era: {palavra}")
jogarForca()
```

```python
#Exercicio 31 - Ordenação por Bolha (Bubble Sort)
#Objetivo - Ordenar uma lista de numeros inteiros em ordem crescente utilizando o algoritmo bubble sort
#Raciocinio - Peco a quantidade de elementos e preencho a lista com as entradas do usuario crio a funcao bubblesort que usa dois laços for para comparar elementos adjacentes e realizar a troca de posicao caso o elemento atual seja maior que o proximo repetindo o processo ate a lista ficar totalmente ordenada
def bubbleSort(lista):
    n = len(lista)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
print(f"\nLista original: {lista}")
listaOrdenada = bubbleSort(lista)
print(f"Lista ordenada: {listaOrdenada}")
```

```python
#Exercicio 32 - Jogo da velha
#Objetivo Implementar um jogo da velha interativo para dois jogadores no terminal alternando jogadas ate haver um vencedor ou empate
#Raciocinio - Crio funcoes para inicializar e exibir o tabuleiro validar combinacoes vitoriosas e checar se esta cheio no laco principal alterno entre os jogadores x e o leio a posicao escolhida de 1 a 9 e verifico se esta livre atualizando o tabuleiro e encerrando a partida quando houver um ganhador ou der empate
def criarTabuleiro():
    return [" " for _ in range(9)]
def mostrarTabuleiro(tabuleiro):
    print(f"\n {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
def verificarVencedor(tabuleiro):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]
    for a, b, c in combinacoes:
        if tabuleiro[a] != " " and tabuleiro[a] == tabuleiro[b] == tabuleiro[c]:
            return tabuleiro[a]
    return None
def tabuleiroCheio(tabuleiro):
    return " " not in tabuleiro
def jogarVelha():
    tabuleiro = criarTabuleiro()
    jogadorAtual = "X"
    print("=== Jogo da Velha ===")
    print("Posições do tabuleiro:")
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")
    while True:
        mostrarTabuleiro(tabuleiro)
        print(f"\nVez do jogador {jogadorAtual}")
        posicao = input("Escolha uma posição (1-9): ")
        if not posicao.isdigit() or int(posicao) < 1 or int(posicao) > 9:
            print("Posição inválida! Digite um número de 1 a 9.")
            continue
        posicao = int(posicao) - 1
        if tabuleiro[posicao] != " ":
            print("Essa posição já está ocupada")
            continue
        tabuleiro[posicao] = jogadorAtual
        vencedor = verificarVencedor(tabuleiro)
        if vencedor:
            mostrarTabuleiro(tabuleiro)
            print(f"\nO jogador {vencedor} venceu")
            break
        if tabuleiroCheio(tabuleiro):
            mostrarTabuleiro(tabuleiro)
            print("\nEmpate")
            break
        jogadorAtual = "O" if jogadorAtual == "X" else "X"
jogarVelha()
```

```python
#Exercicio 33 - Verificacao de anagramas
#Objetivo - Verificar se duas palavras ou frases informadas pelo usuario sao anagramas entre si
#Raciocinio - Crio a funcao sao_anagramas que remove os espacos em branco e converte os caracteres para minusculo comparando as duas strings ordenadas com sorted se as listas de caracteres forem iguais retorna true e exibo o resultado no final
def sao_anagramas(palavra1, palavra2):
    palavra1 = palavra1.replace(" ", "").lower()
    palavra2 = palavra2.replace(" ", "").lower()
    return sorted(palavra1) == sorted(palavra2)
palavra1 = input("Digite a primeira palavra: ")
palavra2 = input("Digite a segunda palavra: ")
if sao_anagramas(palavra1, palavra2):
    print(f"\n'{palavra1}' e '{palavra2}' são anagramas")
else:
    print(f"\n'{palavra1}' e '{palavra2}' não são anagramas.")
```

```python
#Exercicio 34 - Contador de vogais
#Objetivo - contar a quantidade total de vogais presentes em uma frase ou palavra digitada pelo usuario
#Raciocinio - crio a funcao contarvogais que define uma string com todas as vogais maiusculas e minusculas depois percorro cada caractere do texto fornecido incrementando o contador sempre que encontrar uma vogal e retorno o total para exibicao
def contarVogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for caractere in texto:
        if caractere in vogais:
            contador += 1
    return contador
texto = input("Digite uma frase ou palavra: ")
totalVogais = contarVogais(texto)
print(f"\nA string '{texto}' tem {totalVogais} vogal(is).")
```

```python
#Exercicio 35 - Numeros Fibonacci Até N
#Objetivo - Gerar e exibir os primeiros n termos da sequencia de fibonacci conforme a quantidade solicitada pelo usuario
#Raciocinio - Crio a funcao fibonaccintermos inicializando as variaveis a e b com 0 e 1 e uso um laco for para repetir n vezes a cada passo adiciono o valor atual de a na lista e atualizo a e b onde a recebe b e b recebe a soma dos dois anteriores exibindo a sequencia completa no final
def fibonacciNTermos(n):
    sequencia = []
    a, b = 0, 1
    for _ in range(n):
        sequencia.append(a)
        a, b = b, a + b
    return sequencia
n = int(input("Quantos termos da sequência de Fibonacci você quer gerar? "))
resultado = fibonacciNTermos(n)
print(f"\nOs primeiros {n} termos de Fibonacci: {resultado}")
```

```python
#Exercicio 36 - Gerador de senhas
#Objetivos - Gerar uma senha aleatoria com o tamanho de caracteres especificado pelo usuario combinando letras e numeros
#Raciocinio - Importo as bibliotecas random e string crio a funcao gerarsenha que combina letras maiusculas minusculas e digitos numericos e uso um laco for para sortear com random.choice um caractere por vez ate atingir o tamanho desejado retornando a senha final para ser exibida
import random
import string
def gerarSenha(tamanho):
    caracteres = string.ascii_letters + string.digits
    senha = ""
    for _ in range(tamanho):
        senha += random.choice(caracteres)
    return senha
tamanho = int(input("Quantos caracteres a senha deve ter? "))
senhaGerada = gerarSenha(tamanho)
print(f"\nSenha gerada: {senhaGerada}")
```

```python
#Exercicio 37 -  Funções Recursivas (Fatorial)
#Objetivo - Calcular o fatorial de um numero inteiro informado pelo usuario utilizando o conceito de recursividade
#Raciocinio - Crio a funcao fatorial que estabelece o caso base onde o fatorial de 0 ou 1 e 1 e no passo recursivo multiplica o numero atual pela chamada da propria funcao decrementada em 1 depois peco o numero ao usuario e exibo o resultado obtido
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    return n * fatorial(n - 1)
numero = int(input("Digite um número para calcular o fatorial: "))
resultado = fatorial(numero)
print(f"\nO fatorial de {numero} é {resultado}")
```

```python
#Exercicio 40 - Lista de compras
#Objetivo - Gerenciar uma lista de compras interativa no terminal permitindo adicionar remover e visualizar itens ate o usuario optar por sair
#Raciocinio - Crio funcoes especificas para adicionar remover e listar os elementos armazenados em uma lista principal depois uso um laco while infinito para exibir um menu interativo ler a opcao do usuario e executar a acao correspondente encerrando o programa quando for selecionada a opcao de sair
def adicionarItem(lista, item):
    lista.append(item)
    print(f"\n'{item}' foi adicionado à lista.")
def removerItem(lista, item):
    if item in lista:
        lista.remove(item)
        print(f"\n'{item}' foi removido da lista.")
    else:
        print(f"\n'{item}' não está na lista.")
def listarItens(lista):
    if not lista:
        print("\nA lista de compras está vazia.")
    else:
        print("\nLista de compras:")
        for i, item in enumerate(lista, 1):
            print(f"{i}. {item}")
listaCompras = []
while True:
    print("\n1. Adicionar item")
    print("2. Remover item")
    print("3. Listar itens")
    print("4. Sair")
    opcao = input("Escolha uma opção: ")
    if opcao == "1":
        item = input("Digite o item a adicionar: ")
        adicionarItem(listaCompras, item)
    elif opcao == "2":
        item = input("Digite o item a remover: ")
        removerItem(listaCompras, item)
    elif opcao == "3":
        listarItens(listaCompras)
    elif opcao == "4":
        print("\nAté logo")
        break
    else:
        print("\nOpção inválida")
```

```python
#Exercicio 41 - Remover duplicadas
#Objetivo - Remover elementos repetidos de uma lista informada pelo usuario mantendo apenas a primeira ocorrencia de cada numero
#Raciocinio - Peco a quantidade de elementos e preencho a lista com as entradas do usuario crio a funcao removerduplicatas que percorre os itens da lista original e adiciona numa nova lista apenas aqueles que ainda nao foram inseridos garantindo elementos unicos na ordem em que apareceram
def removerDuplicatas(lista):
    semDuplicatas = []
    for numero in lista:
        if numero not in semDuplicatas:
            semDuplicatas.append(numero)
    return semDuplicatas
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
listaSemDuplicatas = removerDuplicatas(lista)
print(f"\nLista original: {lista}")
print(f"Lista sem duplicatas: {listaSemDuplicatas}")
```

```python
#Exercicio 42 - Numeros primos ate N
#Objetivo - Encontrar e exibir todos os numeros primos no intervalo de 2 ate um numero n informado pelo usuario
#Raciocinio - Crio a funcao ehprimo que valida se um numero e primo testando se ele e menor que 2 ou se possui divisores ate a sua raiz quadrada crio a funcao primosaten que percorre os numeros de 2 ate n filtrando apenas os primos e armazenando em uma lista exibindo o resultado no final
def ehPrimo(numero):
    if numero < 2:
        return False
    for i in range(2, int(numero ** 0.5) + 1):
        if numero % i == 0:
            return False
    return True
def primosAteN(n):
    primos = []
    for numero in range(2, n + 1):
        if ehPrimo(numero):
            primos.append(numero)
    return primos
n = int(input("Digite até qual número gerar os primos: "))
resultado = primosAteN(n)
print(f"\nNúmeros primos até {n}: {resultado}")
```

```python
#exercicio 43 - Conversor de unidades
#Objetivo - Converter um valor de medida de comprimento entre diferentes unidades (mm, cm, m, km) fornecidas pelo usuario
#Raciocinio - Crio a funcao convertercomprimento que utiliza um dicionario de fatores de conversao tendo o metro como base converto primeiro o valor de origem para metros e em seguida divido pelo fator da unidade de destino retornando o valor final calculado para exibicao
def converterComprimento(valor, unidadeOrigem, unidadeDestino):
    paraMetros = {
        "mm": 0.001,
        "cm": 0.01,
        "m": 1,
        "km": 1000
    }
    valorEmMetros = valor * paraMetros[unidadeOrigem]
    valorConvertido = valorEmMetros / paraMetros[unidadeDestino]
    return valorConvertido
valor = float(input("Digite o valor a converter: "))
unidadeOrigem = input("Digite a unidade de origem (mm, cm, m, km): ")
unidadeDestino = input("Digite a unidade de destino (mm, cm, m, km): ")
resultado = converterComprimento(valor, unidadeOrigem, unidadeDestino)
print(f"\n{valor} {unidadeOrigem} equivale a {resultado} {unidadeDestino}")
```

```python
#Exercicio 45 - Par ou Ímpar em Lista
#Objetivo - Separar os numeros de uma lista fornecida pelo usuario em duas listas distintas uma para valores pares e outra para impares
#Raciocinio - Peco a quantidade de elementos e preencho a lista inicial com as entradas do usuario crio a funcao classificarparesimpares que percorre a lista testando se cada numero e divisivel por 2 (% 2 == 0) adicionando os pares na lista correspondente e os demais na de impares retornando ambas para exibicao
def classificarParesImpares(lista):
    pares = []
    impares = []
    for numero in lista:
        if numero % 2 == 0:
            pares.append(numero)
        else:
            impares.append(numero)
    return pares, impares
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
pares, impares = classificarParesImpares(lista)
print(f"\nLista original: {lista}")
print(f"Pares: {pares}")
print(f"Ímpares: {impares}")
```

```python
#Exercicio 46 - Calcular media e desvio padrao
#Objetivo - calcular e exibir a media aritmetica e o desvio padrao populacional de uma lista de numeros fornecidos pelo usuario
#Raciocinio - crio a funcao calcularmedia para obter a media dividindo a soma dos elementos pela quantidade total em seguida crio calculardesviopadrao que calcula a variancia dividindo a soma das diferencas quadraticas em relacao a media pelo total de itens e aplica a raiz quadrada elevando a 0.5 por fim leio os numeros do usuario e exibo a lista com os resultados obtidos
def calcularMedia(lista):
    return sum(lista) / len(lista)
def calcularDesvioPadrao(lista):
    media = calcularMedia(lista)
    somaQuadrados = sum((numero - media) ** 2 for numero in lista)
    variancia = somaQuadrados / len(lista)
    return variancia ** 0.5
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = float(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
media = calcularMedia(lista)
desvioPadrao = calcularDesvioPadrao(lista)
print(f"\nLista: {lista}")
print(f"Média: {media}")
print(f"Desvio padrão: {desvioPadrao}")
```

```python
#Exercicio 47 - Contar numeros em lista
#Objetvo - contar a frequencia com que cada numero aparece em uma lista digitada pelo usuario
#Raciocinio - peco a quantidade de elementos e preencho a lista com as entradas do usuario crio a funcao contarnumeros que utiliza um dicionario para mapear cada valor se o numero ja existe no dicionario incremento sua contagem caso contrario inicializo com 1 por fim percorrro as chaves e valores do dicionario para exibir a quantidade de ocorrencias de cada numero
def contarNumeros(lista):
    contagem = {}
    for numero in lista:
        if numero in contagem:
            contagem[numero] += 1
        else:
            contagem[numero] = 1
    return contagem
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
resultado = contarNumeros(lista)
print(f"\nLista: {lista}")
print("Contagem de cada número:")
for numero, vezes in resultado.items():
    print(f"{numero} aparece {vezes} vez(es)")
```

```python
#Exercicio 48 - Ordenacao por Insercao (Insertion Sort)
#Objetivo - Ordenar uma lista de numeros em ordem crescente utilizando o algoritmo insertion sort
#raciocinio - Peco a quantidade de elementos e preencho a lista inicial crio a funcao insertionsort que percorre a lista a partir do segundo elemento i comparando a chave atual com os elementos anteriores enquanto houver valores maiores a esquerda desloco esses elementos para a direita e insiro o valor atual na posicao correta exibindo a lista ordenada ao final
def insertionSort(lista):
    for i in range(1, len(lista)):
        atual = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > atual:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = atual
    return lista
quantidade = int(input("Quantos números você quer digitar? "))
lista = []
for i in range(quantidade):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)
print(f"\nLista original: {lista}")
listaOrdenada = insertionSort(lista)
print(f"Lista ordenada: {listaOrdenada}")
```

```python
#Exercicio 49 - Soma dos digitos
#Objetivo - Calcular a soma de todos os digitos individuais de um numero inteiro fornecido pelo usuario
#Raciocinio - Crio a funcao somarDigitos que converte o numero para seu valor absoluto com abs tratando valores negativos em seguida uso um laco while para extrair o ultimo digito com o operador de resto % 10 acumular na variavel soma e remover o ultimo digito usando a divisao inteira // 10 ate que o numero seja zero exibindo o resultado ao final
def somarDigitos(numero):
    soma = 0
    numero = abs(numero)
    while numero > 0:
        soma += numero % 10
        numero //= 10
    return soma
numero = int(input("Digite um número inteiro: "))
resultado = somarDigitos(numero)
print(f"\nA soma dos dígitos de {numero} é {resultado}")
```

```python
#Exercicio 50 - Calculo de frequencia de letras
#Objetivo - Calcular a frequencia de ocorrencia de cada letra em uma frase ou palavra fornecida pelo usuario ignorando espacos e pontuacoes
#Raciocinio - Crio a funcao calcularfrequencia que converte todo o texto para minusculas e percorre cada caractere filtrando apenas letras com isalpha caso a letra ja exista no dicionario incremento o contador se nao a inicializo com 1 por fim percorro o dicionario para exibir a quantidade de vezes que cada letra aparece
def calcularFrequencia(texto):
    frequencia = {}
    for caractere in texto.lower():
        if caractere.isalpha():
            if caractere in frequencia:
                frequencia[caractere] += 1
            else:
                frequencia[caractere] = 1
    return frequencia
texto = input("Digite uma frase ou palavra: ")
resultado = calcularFrequencia(texto)
print(f"\nFrequência de cada letra:")
for letra, vezes in resultado.items():
    print(f"{letra}: {vezes}")
```
