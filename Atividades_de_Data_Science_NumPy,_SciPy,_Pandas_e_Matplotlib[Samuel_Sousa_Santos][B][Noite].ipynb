```python
#Exercıcios de NumPy
import numpy as np
#Raciocinio - Uso np.random.randint(baixo, alto, tamanho) para gerar valores inteiros dentro de um intervalo. O limite superior em randint e exclusivo, por isso uso 101 para incluir o 100.
#Gerando array de 50 inteiros entre 10 e 100
array = np.random.randint(10, 101, 50)
print("Array de 50 numeros aleatorios entre 10 e 100:")
print(array)
```

```python
import numpy as np
# Raciocinio - Gero os números de 1 a 16 com np.arange e uso .reshape(4, 4) para organizá-los em uma matriz 4x4. A última linha é acessada com índice -1.
# Criando a matriz 4x4 e extraindo a última linha
matriz = np.arange(1, 17).reshape(4, 4)
print("Matriz 4x4:")
print(matriz)
ultima_linha = matriz[-1]
print("\nÚltima linha:", ultima_linha)
```

```python
import numpy as np
# Raciocinio - Crio um array de 15 zeros com np.zeros(15). Como a indexação começa em 0, a 5ª posição corresponde ao índice 4 e a 10ª ao índice 9; uso slicing [4:10] para substituir esse intervalo por 1 de uma vez.
# Criando o array e substituindo o intervalo
array = np.zeros(15)
array[4:10] = 1
print("Array com a 5ª à 10ª posição substituída por 1:")
print(array)
```

```python
import numpy as np
# Raciocinio - Gero 100 inteiros aleatorios e uso np.median() para a mediana (valor central quando os dados estao ordenados). NumPy nao tem uma funcao nativa de moda, entao uso np.bincount() combinado com np.argmax(), que conta a frequência de cada valor e retorna o índice (valor) mais frequente.
# Gerando 100 inteiros e calculando mediana e moda
array = np.random.randint(0, 100, 100)
mediana = np.median(array)
contagens = np.bincount(array)
moda = np.argmax(contagens)
print("Array:", array)
print("\nMediana:", mediana)
print("Moda:", moda)
```

```python
import numpy as np
# Raciocinio - Gero um array de 20 elementos entre 1 e 100 e uso indexação booleana: array % 3 == 0 cria uma máscara True/False que identifica os múltiplos de 3, e a atribuição array[mascara] = -3 substitui só esses valores, sem precisar de laço explícito.
# Gerando o array e substituindo multiplos de 3 por -3
array = np.random.randint(1, 101, 20)
print("Array original:", array)
array[array % 3 == 0] = -3
print("Array com multiplos de 3 substituidos por -3:", array)
```

```python
import numpy as np
# Raciocinio - Gero uma matriz 5x5 com np.random.rand() (valores entre 0 e 1) e uso np.sum(matriz, axis=0). O parametro axis=0 indica que a soma deve ser feita ao longo das linhas (ou seja, coluna por coluna).
# Gerando a matriz 5x5 e somando por colunas
matriz = np.random.rand(5, 5)
print("Matriz:")
print(matriz)
soma_colunas = np.sum(matriz, axis=0)
print("\nSoma de cada coluna:", soma_colunas)
```

```python
import numpy as np
# Raciocinio - Uso np.corrcoef(a, b), que retorna uma matriz de correlacao 2x2 — a posicao [0, 1] (ou [1, 0]) contem o coeficiente de correlacao entre os dois arrays, que varia de -1 (correlacao negativa perfeita) a 1 (correlacao positiva perfeita).
# Gerando dois arrays aleatorios e calculando a correlacao
a = np.random.rand(20)
b = np.random.rand(20)
correlacao = np.corrcoef(a, b)[0, 1]
print("Correlacao entre os arrays:", correlacao)
```

```python
import numpy as np
# Raciocinio - Gero uma matriz 3x3 de inteiros aleatorios e uso o operador ** 2, que no NumPy e aplicado elemento a elemento (operacao vetorizada) — nao e necessario laco para percorrer a matriz manualmente.
# Gerando a matriz 3x3 e elevando seus elementos ao quadrado
matriz = np.random.randint(1, 20, (3, 3))
print("Matriz original:")
print(matriz)
matriz_quadrado = matriz ** 2
print("\nMatriz elevada ao quadrado:")
print(matriz_quadrado)
```
