```python
#Exercicio 19
import numpy as np
# Raciocinio - O cosseno do angulo entre dois vetores e dado por cos(theta) = (a.b) / (|a| |b|), onde a.b e o produto escalar (np.dot) e |a|, |b| sao as normas (magnitudes) dos vetores (np.linalg.norm). Depois de calcular o cosseno, uso np.arccos() para obter o angulo em radianos, e converto para graus com np.degrees() para facilitar a leitura.
# Gerando dois arrays e calculando o angulo entre eles
arrayA = np.random.randint(1, 20, 20)
arrayB = np.random.randint(1, 20, 20)
cosTheta = np.dot(arrayA, arrayB) / (np.linalg.norm(arrayA) * np.linalg.norm(arrayB))
anguloRadianos = np.arccos(cosTheta)
anguloGraus = np.degrees(anguloRadianos)
print("Angulo entre os arrays:")
print(f"{anguloGraus:.2f} graus")
```

```python
#Exercícios 1 de SciPy
import numpy as np
from scipy import integrate
# Raciocinio - Uso scipy.integrate.quad(), que calcula integrais definidas numericamente. Passo a função f(x) = x² * sin(x), o limite inferior (0) e o superior (π, obtido de np.pi).
# Definindo a função e realizando a integração
def f(x):
    return (x**2) * np.sin(x)
resultado, erro = integrate.quad(f, 0, np.pi)
print("Integral de x² sin(x) de 0 a π:")
print(f"Resultado: {resultado:.6f}")
print(f"Erro estimado: {erro:.2e}")
```

```python
#Exercicio 2
import numpy as np
from scipy.integrate import solve_ivp
# Raciocinio - Uso scipy.integrate.solve_ivp() para resolver a equação diferencial ordinária numericamente. Defino a função que representa dy/dx em termos de x e y, passo o intervalo [0, 10] e a condição inicial y(0) = 1. t_eval define os pontos onde quero ver a solução.
# Definindo a equação diferencial e resolvendo o PVI (Problema de Valor Inicial)
def eq_diferencial(x, y):
    return x - y
solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 50))
print("Valores de x:")
print(solucao.t)
print("\nValores de y:")
print(solucao.y[0])
```

```python
#Exercicio 3
import numpy as np
# Raciocinio - Como essa é uma função polinomial, uso np.roots(), que recebe os coeficientes do polinômio (do maior para o menor grau) e retorna todas as raízes de uma vez.
# Coeficientes de x^4 - 3x^3 + 0x^2 + 0x + 2
coeficientes = [1, -3, 0, 0, 2]
raizes = np.roots(coeficientes)
print("Raízes de f(x) = x⁴ - 3x³ + 2:")
for r in raizes:
    if np.isreal(r):
        print(f"{r.real:.4f}")
    else:
        print(f"{r:.4f}")
```

```python
#Exercicio 4
import numpy as np
from scipy.optimize import minimize
# Raciocinio - O SciPy só tem funções de minimização prontas (como optimize.minimize), então para maximizar f(x) minimizamos o seu negativo (-f(x)). O ponto x que minimiza -f(x) é o mesmo que maximiza f(x).
def f(x):
    return -(x**2) + 4*x + 1

def f_negativa(x):
    return -f(x)
# Chute inicial em x = 0
resultado = minimize(f_negativa, x0=0)
x_otimo = resultado.x[0]
y_max = f(x_otimo)
print("x que maximiza f(x):")
print(f"{x_otimo:.4f}")
print("\nValor máximo de f(x):")
print(f"{y_max:.4f}")
```

```python
#Exercicio 5
import numpy as np
from scipy.fft import fft, fftfreq
# Raciocinio - Gero o sinal combinando as duas ondas no tempo (t de 0 a 1s com 500 amostras) e uso scipy.fft.fft() para converter do domínio do tempo para o domínio da frequência. Também uso fftfreq para obter o eixo de frequências associado.
N = 500
t = np.linspace(0, 1, N, endpoint=False)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
# Cálculo da FFT e das frequências correspondentes
frequencias = fft(sinal)
eixo_freq = fftfreq(N, d=1/N)
print("Transformada de Fourier (primeiros 10 coeficientes complexos):")
print(frequencias[:10])
print("\nMagnitude dos primeiros 10 coeficientes:")
print(np.abs(frequencias[:10]))
```

```python
#Exercicio 6
import numpy as np
from scipy import linalg
# Raciocinio - Gero uma matriz 5x5 aleatória e uso scipy.linalg.inv(), que calcula a matriz inversa (aquela que, multiplicada pela original, resulta na matriz identidade).
# Gerando a matriz 5x5 aleatória e calculando a inversa
matrizOriginal = np.random.rand(5, 5)
matrizInversa = linalg.inv(matrizOriginal)
print("Matriz original:")
print(matrizOriginal)
print("\nMatriz inversa:")
print(matrizInversa)
# Validação: matrizOriginal @ matrizInversa deve resultar na matriz Identidade 5x5
print("\nVerificação (Matriz x Inversa ≈ Identidade):")
print(np.round(matrizOriginal @ matrizInversa, 2))
```

```python
#Exercicio 8
import numpy as np
from scipy import linalg
# Raciocinio - Escrevo o sistema na forma matricial Ax = B, onde A contém os coeficientes das incógnitas e B os resultados de cada equação, e uso scipy.linalg.solve() para resolver diretamente.
# Definindo a matriz de coeficientes (A) e o vetor de termos independentes (B)
A = np.array([[3, 2], [1, 4]])
B = np.array([5, 6])
# Resolvendo o sistema linear
solucao = linalg.solve(A, B)
print("Solução do sistema [x, y]:")
print(solucao)
print(f"\nx = {solucao[0]:.4f}")
print(f"y = {solucao[1]:.4f}")
```

```python
#Exercicio 9
import numpy as np
from scipy.optimize import curve_fit
# Raciocinio - Defino a forma da curva desejada (y = ax² + bx + c) como uma funcao e uso scipy.optimize.curve_fit(), que encontra os parametros a, b e c que melhor ajustam essa curva aos pontos fornecidos, minimizando o erro entre a curva e os dados reais.
# Definindo os pontos conhecidos
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])
# Definindo o modelo da parabola (funcao quadratica)
def parabola(x, a, b, c):
    return a * x**2 + b * x + c
# Ajustando os parametros a, b e c aos dados
parametros, _ = curve_fit(parabola, x, y)
print("Parametros ajustados [a, b, c]:")
print([round(p, 4) for p in parametros])
print(f"\nEquacao da curva ajustada: y = {parametros[0]:.4f}x² + {parametros[1]:.4f}x + {parametros[2]:.4f}")
```

```python
#Exercicio 10
import numpy as np
from scipy import integrate
# Raciocinio - Uso scipy.integrate.dblquad(), que calcula integrais duplas. A ordem dos argumentos e importante: passo a funcao f(y, x), os limites da variavel externa x (0 a 2) e os limites da variavel interna y como funcoes lambda (0 a 1).
def f(y, x):
    return x * y + x**2
resultado, erro = integrate.dblquad(f, 0, 2, lambda x: 0, lambda x: 1)
print("Integral dupla:")
print(f"Resultado: {resultado:.4f}")
print("\nErro estimado:")
print(f"{erro:.2e}")
```

```python
#Exercicio 11
import numpy as np
from scipy import linalg
# Raciocinio - Uso scipy.linalg.eig(), que retorna dois resultados: os autovalores (escalares que indicam "quanto" a matriz estica cada direcao) e os autovetores correspondentes (as direcoes que nao mudam de orientacao sob a transformacao da matriz).
# Criando matriz 4x4 de inteiros
matriz = np.random.randint(1, 10, (4, 4)).astype(float)
# Calculando autovalores e autovetores
autovalores, autovetores = linalg.eig(matriz)
print("Matriz:")
print(matriz)
print("\nAutovalores:")
print(np.round(autovalores, 4))
print("\nAutovetores (colunas correspondentes):")
print(np.round(autovetores, 4))
```

```python
#Exercicio 14
import numpy as np
from scipy.optimize import minimize
# Raciocinio - Como a funcao tem duas variaveis, passo um unico array [x, y] para scipy.optimize.minimize(), que otimiza as duas variaveis simultaneamente a partir de um ponto inicial (aqui, [0, 0]).
def funcao(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2
# Estimativa inicial em (0, 0)
resultado = minimize(funcao, [0, 0])
print("Minimo encontrado em [x, y]:")
print(np.round(resultado.x, 4))
print("\nValor minimo de f(x, y):")
print(f"{resultado.fun:.4f}")
```

```python
#Exercicio 15
import numpy as np
from scipy.fft import ifft
# Raciocinio - Gero um sinal complexo aleatorio (parte real e parte imaginaria) e uso scipy.fft.ifft(), que faz o processo inverso da FFT: converte um sinal do dominio da frequencia de volta para o dominio do tempo.
# Gerando sinal complexo no dominio da frequencia (20 pontos)
sinal_complexo = np.random.rand(20) + 1j * np.random.rand(20)
# Aplicando a IFFT para reconstruir no dominio do tempo
sinal_tempo = ifft(sinal_complexo)
print("Sinal complexo original (primeiros 5):")
print(np.round(sinal_complexo[:5], 4))
print("\nApos transformada inversa (primeiros 5):")
print(np.round(sinal_tempo[:5], 4))
```

```python
#Exercicio 16
import numpy as np
from scipy.integrate import solve_ivp
# Raciocinio - solve_ivp so resolve equacoes de 1a ordem, entao transformo a equacao de 2a ordem em um sistema de duas equacoes de 1a ordem: chamo y' = v, o que da v' = -y. A condicao inicial vira um vetor [y(0), y'(0)] = [1, 0].
def sistema(x, estado):
    y, v = estado
    dydx = v
    dvdx = -y
    return [dydx, dvdx]
# Resolvendo o PVI de x=0 ate x=10 com 100 pontos de avaliacao
solucao = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100))
print("x (primeiros 10 pontos):")
print(np.round(solucao.t[:10], 4))
print("\ny = cos(x) (primeiros 10 pontos):")
print(np.round(solucao.y[0][:10], 4))
```

```python
#Exercicio 17
import numpy as np
from scipy.optimize import brentq
# Raciocinio - Embora 27 ** (1/3) ja resolvesse diretamente, uso scipy.optimize.brentq() para demonstrar a busca de raiz: procuro o valor de x onde x³ - 27 = 0, dentro de um intervalo onde sei que a raiz existe (entre 0 e 10).
# Definindo a funcao f(x) = x³ - 27
def f(x):
    return x**3 - 27
# Buscando a raiz no intervalo [0, 10]
raiz = brentq(f, 0, 10)
print("Raiz cubica de 27:")
print(f"{raiz:.4f}")
```

```python
#Exercicio 18
import numpy as np
from scipy.interpolate import CubicSpline
# Raciocinio - Gero os pontos conhecidos de f(x) = x² e uso scipy.interpolate.CubicSpline, que ajusta uma spline cubica (curvas suaves conectadas nos pontos) — mais preciso que interpolacao linear para funcoes curvas.
# Pontos de origem
x = np.array([1, 2, 3, 4, 5])
y = x**2
# Criando a spline cubica
spline = CubicSpline(x, y)
# Avaliando em 10 pontos no intervalo [1, 5]
x_novos = np.linspace(1, 5, 10)
y_novos = spline(x_novos)
print("Valores x interpolados:")
print(np.round(x_novos, 4))
print("\nValores y interpolados com spline cubica:")
print(np.round(y_novos, 4))
```

```python
#Exercicio 19
import numpy as np
from scipy.fft import fft, fftfreq
# Raciocinio - Monto um sinal combinando multiplas componentes senoidais e uso scipy.fft.fft() para decompor esse sinal discreto em suas frequencias constituintes.
# Criando o vetor de tempo de 0 a 1 segundo (200 pontos)
t = np.linspace(0, 1, 200, endpoint=False)
# Sinal composto por uma senoide de 5 Hz e uma cossenoide de 10 Hz
sinal = np.sin(2 * np.pi * 5 * t) + np.cos(2 * np.pi * 10 * t)
# Calculando a FFT (Transformada Rapida de Fourier)
frequencias = fft(sinal)
print("Transformada de Fourier (primeiros 10 valores complexos):")
print(np.round(frequencias[:10], 4))
print("\nMagnitude dos 10 primeiros componentes de frequencia:")
print(np.round(np.abs(frequencias[:10]), 4))
```

```python
#Exercicio 20
import numpy as np
from scipy import linalg
# Raciocinio - Mesma logica do exercicio 8, mas agora com um sistema 3x3: monto a matriz de coeficientes A (3 equacoes, 3 incognitas) e o vetor de resultados B, e resolvo com scipy.linalg.solve().
# Sistema:
#  2x +  y -  z = 3
#   x -  y + 2z = 1
#  3x + 2y -  z = 7
A = np.array([
    [2,  1, -1],
    [1, -1,  2],
    [3,  2, -1]
])
B = np.array([3, 1, 7])
# Resolvendo A * x = B
solucao = linalg.solve(A, B)
print("Solucao [x, y, z]:")
print(np.round(solucao, 4))
```

```python
#Exercicio 8
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Ploto a linha principal com plt.plot() e uso plt.fill_between() para sombrear a área entre y - margem e y + margem, simulando um intervalo de confiança ao redor da curva.
# Definindo o intervalo de x
x = np.linspace(0, 10, 100)
y = np.sin(x)
margem = 0.2
# Plotando a curva principal
plt.plot(x, y, color='blue', label='sin(x)')
# Adicionando o sombreamento (intervalo de confiança)
plt.fill_between(x, y - margem, y + margem, color='blue', alpha=0.2, label='Intervalo de confiança')
# Personalizando o gráfico
plt.title('Gráfico com intervalo de confiança')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 10
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Chamo plt.plot() três vezes na mesma figura (o Matplotlib acumula os plots até o plt.show()), cada chamada com uma cor (color) e estilo de linha (linestyle) diferente.
# Definindo o intervalo de x
x = np.linspace(0, 10, 100)
# Plotando as três linhas com cores e estilos de linha diferentes
plt.plot(x, np.sin(x), color='blue', linestyle='-', label='sin(x)')
plt.plot(x, np.cos(x), color='red', linestyle='--', label='cos(x)')
plt.plot(x, np.sin(x) * np.cos(x), color='green', linestyle=':', label='sin(x)*cos(x)')
# Personalizando o gráfico
plt.title('Três funções no mesmo gráfico')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 11
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Uso o parâmetro bottom em plt.bar() para empilhar as barras: cada grupo novo começa de onde o grupo anterior parou (soma acumulada), em vez de começar do zero.
# Dados de origem
categorias = ['A', 'B', 'C']
grupo1 = np.array([10, 15, 8])
grupo2 = np.array([5, 7, 9])
grupo3 = np.array([3, 4, 6])
# Criando o gráfico de barras empilhadas
plt.bar(categorias, grupo1, label='Grupo 1')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')
plt.bar(categorias, grupo3, bottom=grupo1 + grupo2, label='Grupo 3')
# Personalizando o gráfico
plt.title('Barras Empilhadas')
plt.xlabel('Categoria')
plt.ylabel('Valor')
plt.legend()
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 12
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Ploto a parábola normalmente e uso plt.xlim() para restringir a área visível do eixo x, mesmo que os dados calculados cubram uma faixa maior.
# Definindo o intervalo de x
x = np.linspace(-10, 10, 200)
y = x**2
# Criando o gráfico
plt.plot(x, y)
# Restringindo os limites de exibição no eixo x
plt.xlim([-5, 5])
# Personalizando o gráfico
plt.title('Função Quadrática (zoom em x: -5 a 5)')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 13
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Uso o módulo mpl_toolkits.mplot3d, criando os eixos com projection='3d'. Depois, ax.scatter() funciona como o plt.scatter() normal, mas recebendo três coordenadas (x, y, z) em vez de duas.
# Gerando os dados aleatórios
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)
# Criando a figura e adicionando os eixos 3D
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
# Plotando a dispersão tridimensional
ax.scatter(x, y, z)
# Definindo títulos e rótulos dos três eixos
ax.set_title('Dispersão 3D')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 16
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Uso plt.text(x, y, texto) para adicionar uma anotação em uma posição específica do gráfico — calculo o valor de y correspondente a x=5 para posicionar o texto exatamente sobre a curva.
# Definindo o intervalo de x
x = np.linspace(0, 10, 100)
y = np.sin(x)
# Criando o gráfico da curva principal
plt.plot(x, y)
# Personalizando títulos e rótulos
plt.title('Função Seno com Anotação')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
# Calculando a posição no ponto x = 5 e adicionando o destaque e o texto
y_em_5 = np.sin(5)
plt.text(5, y_em_5, '  x = 5', fontsize=10, color='red')
plt.scatter([5], [y_em_5], color='red')
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 17
import matplotlib.pyplot as plt
import numpy as np
# Raciocínio: Uso plt.savefig('nome.pdf') antes de plt.show() (chamar depois pode gerar um arquivo em branco, pois o show() limpa a figura da memória em alguns ambientes) para salvar o gráfico atual em disco no formato PDF.
# Definindo o intervalo de x
x = np.linspace(0, 10, 100)
y = np.sin(x)
# Criando o gráfico
plt.plot(x, y)
# Personalizando o gráfico
plt.title('Gráfico salvo em PDF')
# Salvando a figura ANTES do plt.show()
plt.savefig('grafico_seno.pdf')
print("Gráfico salvo como 'grafico_seno.pdf'")
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 18
import matplotlib.pyplot as plt
# Raciocínio: Gero uma lista de cores condicionalmente: verde para valores positivos, vermelho para negativos, usando uma list comprehension que testa cada valor antes de plotar.
# Dados de origem
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [10, -5, 8, -3, 6]
# Definindo as cores com base no valor
cores = ['green' if v >= 0 else 'red' for v in valores]
# Criando o gráfico de barras
plt.bar(categorias, valores, color=cores)
# Adicionando linha de referência no y=0 para destacar a transição
plt.axhline(0, color='black', linewidth=0.8)
# Personalizando o gráfico
plt.title('Valores Positivos e Negativos')
plt.xlabel('Categoria')
plt.ylabel('Valor')
# Exibindo o gráfico
plt.show()
```

```python
#Exercicio 19
import matplotlib.pyplot as plt
import numpy as np
# Raciocinio: Ploto duas curvas na mesma figura, cada uma com o parâmetro label definido, e uso plt.legend() para exibir a caixa de legenda que identifica qual cor/linha corresponde a qual série.
# Definindo o intervalo de x
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
# Plotando as duas series informando os rótulos (labels)
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
# Personalizando o grafico
plt.title('Duas Séries com Legenda')
# Exibindo a legenda para identificar as series
plt.legend()
# Exibindo o grafico
plt.show()
```

```python
#Exercicio 20
import matplotlib.pyplot as plt
import numpy as np
# Raciocinio: Uso plt.xticks() e plt.yticks() com np.arange() para definir manualmente onde as linhas de grade (e os números do eixo) devem aparecer, em vez de deixar o Matplotlib escolher automaticamente.

# Definindo o intervalo de x e calculando y
x = np.linspace(-10, 10, 100)
y = x**2
# Criando o grafico
plt.plot(x, y)
# Definindo o espaçamento personalizado para os eixos X e Y
plt.xticks(np.arange(-10, 11, 2))
plt.yticks(np.arange(0, 101, 10))
# Personalizando o grafico
plt.title('y = x² com grade personalizada')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
# Exibindo o grafico
plt.show()
```
