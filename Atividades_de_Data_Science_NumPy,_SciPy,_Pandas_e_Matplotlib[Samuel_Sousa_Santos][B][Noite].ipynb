```python
import numpy as np
# Raciocinio - Gero 100 inteiros aleatorios e uso np.median() para a mediana (valor central quando os dados estao ordenados). NumPy nao tem uma funcao nativa de moda, entao uso np.bincount() combinado com np.argmax(), que conta a frequencia de cada valor e retorna o indice (valor) mais frequente.
# Gerando 100 inteiros e calculando mediana e moda
arrayNumeros = np.random.randint(0, 100, 100)
medianaArray = np.median(arrayNumeros)
contagensValores = np.bincount(arrayNumeros)
modaArray = np.argmax(contagensValores)
print("Array:", arrayNumeros)
print("\nMediana:", medianaArray)
print("Moda:", modaArray)
```

```python
import numpy as np
# Raciocinio - Uso np.corrcoef(arrayA, arrayB), que retorna uma matriz de correlacao 2x2 — a posicao [0, 1] (ou [1, 0]) contem o coeficiente de correlacao entre os dois arrays, que varia de -1 (correlacao negativa perfeita) a 1 (correlacao positiva perfeita).
# Gerando dois arrays aleatorios e calculando a correlacao
arrayA = np.random.rand(20)
arrayB = np.random.rand(20)
matrizCorrelacao = np.corrcoef(arrayA, arrayB)
coeficienteCorrelacao = matrizCorrelacao[0, 1]
print("Correlacao entre os arrays:", coeficienteCorrelacao)
```

```python
import numpy as np
# Raciocinio - A normalizacao Min-Max transforma qualquer valor x em (x - minimo) / (maximo - minimo), garantindo que o menor valor vire 0 e o maior vire 1. Uso np.set_printoptions(linewidth=100) e imprimo o texto separado do array para evitar quebras e espacamentos indesejados.
# Gerando o array e aplicando a normalizacao Min-Max
np.set_printoptions(linewidth=100)
arrayOriginal = np.random.randint(1, 100, 10).astype(float)
print("Array original:")
print(arrayOriginal)
arrayNormalizado = (arrayOriginal - arrayOriginal.min()) / (arrayOriginal.max() - arrayOriginal.min())
print("\nArray normalizado entre 0 e 1:")
print(np.round(arrayNormalizado, 4))
```

```python
import numpy as np
# Raciocinio - Mesma logica do exercicio 4 (calculo de moda): np.bincount() conta quantas vezes cada valor aparece, e np.argmax() retorna o valor (indice) com maior contagem — ou seja, o mais frequente.
# Gerando o array e encontrando o valor mais frequente
arrayNumeros = np.random.randint(0, 20, 100)
contagensValores = np.bincount(arrayNumeros)
maisRepetido = np.argmax(contagensValores)
print("Array:")
print(arrayNumeros)
print("\nValor mais repetido:", maisRepetido, "- apareceu", contagensValores[maisRepetido], "vezes")
```

```python
import numpy as np
# Raciocinio - O cosseno do angulo entre dois vetores e dado por cos(theta) = (a.b) / (|a| |b|), onde a.b e o produto escalar (np.dot) e |a|, |b| sao as normas (magnitudes) dos vetores (np.linalg.norm). Depois de calcular o cosseno, uso np.arccos() para obter o angulo em radianos, e converto para graus com np.degrees() para facilitar a leitura.
# Gerando dois arrays e calculando o angulo entre eles
arrayA = np.random.randint(1, 20, 20)
arrayB = np.random.randint(1, 20, 20)
cosTheta = np.dot(arrayA, arrayB) / (np.linalg.norm(arrayA) * np.linalg.norm(arrayB))
anguloRadianos = np.arccos(cosTheta)
anguloGraus = np.degrees(anguloRadianos)
print("Angulo entre os arrays:")
print(f"{anguloGraus:.2f} graus")
```

```python
#Exercícios de SciPy -  Integral de f(x) = x² sin(x) de 0 a π
import numpy as np
from scipy import integrate
# Raciocínio - Uso scipy.integrate.quad(), que calcula integrais definidas numericamente. Passo a função f(x) = x² * sin(x), o limite inferior (0) e o superior (π, obtido de np.pi).
# Definindo a função e realizando a integração
def f(x):
    return (x**2) * np.sin(x)
resultado, erro = integrate.quad(f, 0, np.pi)
print("Integral de x² sin(x) de 0 a π:")
print(f"Resultado: {resultado:.6f}")
print(f"Erro estimado: {erro:.2e}")
```

```python
import numpy as np
from scipy.integrate import solve_ivp
# Raciocínio - Uso scipy.integrate.solve_ivp() para resolver a equação diferencial ordinária numericamente. Defino a função que representa dy/dx em termos de x e y, passo o intervalo [0, 10] e a condição inicial y(0) = 1. t_eval define os pontos onde quero ver a solução.
# Definindo a equação diferencial e resolvendo o PVI (Problema de Valor Inicial)
def eq_diferencial(x, y):
    return x - y
solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 50))
print("Valores de x:")
print(solucao.t)
print("\nValores de y:")
print(solucao.y[0])
```

```python
import numpy as np
# Raciocínio - Como essa é uma função polinomial, uso np.roots(), que recebe os coeficientes do polinômio (do maior para o menor grau) e retorna todas as raízes de uma vez.
# Coeficientes de x^4 - 3x^3 + 0x^2 + 0x + 2
coeficientes = [1, -3, 0, 0, 2]
raizes = np.roots(coeficientes)
print("Raízes de f(x) = x⁴ - 3x³ + 2:")
for r in raizes:
    if np.isreal(r):
        print(f"{r.real:.4f}")
    else:
        print(f"{r:.4f}")
```

```python
import numpy as np
from scipy.optimize import minimize
# Raciocínio - O SciPy só tem funções de minimização prontas (como optimize.minimize), então para maximizar f(x) minimizamos o seu negativo (-f(x)). O ponto x que minimiza -f(x) é o mesmo que maximiza f(x).
def f(x):
    return -(x**2) + 4*x + 1

def f_negativa(x):
    return -f(x)
# Chute inicial em x = 0
resultado = minimize(f_negativa, x0=0)
x_otimo = resultado.x[0]
y_max = f(x_otimo)
print("x que maximiza f(x):")
print(f"{x_otimo:.4f}")
print("\nValor máximo de f(x):")
print(f"{y_max:.4f}")
```

```python
import numpy as np
from scipy.fft import fft, fftfreq
# Raciocínio - Gero o sinal combinando as duas ondas no tempo (t de 0 a 1s com 500 amostras) e uso scipy.fft.fft() para converter do domínio do tempo para o domínio da frequência. Também uso fftfreq para obter o eixo de frequências associado.
N = 500
t = np.linspace(0, 1, N, endpoint=False)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
# Cálculo da FFT e das frequências correspondentes
frequencias = fft(sinal)
eixo_freq = fftfreq(N, d=1/N)
print("Transformada de Fourier (primeiros 10 coeficientes complexos):")
print(frequencias[:10])
print("\nMagnitude dos primeiros 10 coeficientes:")
print(np.abs(frequencias[:10]))
```
